dashboard_layout = {
    "data": [
        {
            "type": "bar",
            "x": data["Product Purchased"].value_counts().index.tolist(),
            "y": data["Product Purchased"].value_counts().values.tolist(),
            "name": "Count of Purchases by Product"
        },
        {
            "type": "histogram",
            "x": data["Amount Spent"],
            "name": "Distribution of Amount Spent"
        }
    ],
    "layout": {
        "title": "Sample Sales Data Dashboard",
        "xaxis": {
            "title": "Product Purchased"
        },
        "yaxis": {
            "title": "Count"
        }
    }
}

# Show the dashboard
fig = px.box(data, x="Gender", y="Amount Spent")
fig.update_layout(dashboard_layout["layout"])
fig.show()
